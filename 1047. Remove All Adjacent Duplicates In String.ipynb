#### Solution 1 in python

```python
class Solution:
    def removeDuplicates(self, s: str) -> str:
        st = []
        for i in s:
            if st:
                if i == st[-1]:
                    st.pop()
                else:
                    st.append(i)
            else:
                st.append(i)

        return "".join(st)
```
