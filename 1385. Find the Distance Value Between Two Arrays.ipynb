#### Solution 1 in python


```python
class Solution:
    def findTheDistanceValue(self, arr1: list[int], arr2: list[int], d: int) -> int:
        
        val = 0
        for i in arr1:
            dis = 0
            for j in arr2:
                if abs(i-j)<=d:
                    continue
                else:
                    dis+=1
            
            if dis==len(arr2):
                val+=1
        return val
                 
```
