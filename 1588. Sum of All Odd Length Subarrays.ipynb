#### Solution 1 in python

```python
class Solution:
    def sumOddLengthSubarrays(self, arr: list[int]) -> int:
        s =0
        n = len(arr)
        for i in range(n+1):
            for j in range(i+1, n+1):
                c=arr[i:j]
                d = len(c)
                if d%2!=0:
                    s+=sum(c)
        
        return s
```
