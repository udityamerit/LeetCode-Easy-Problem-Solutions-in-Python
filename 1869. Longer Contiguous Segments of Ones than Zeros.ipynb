#### Solution 1 in python

```python
class Solution:
    def checkZeroOnes(self, s: str) -> bool:
        one = 0
        zero = 0
        max_one = 0
        max_zero = 0
        for i in s:
            if i!='0':
                one+=1
                zero = 0
            else:
                zero+=1
                one = 0
            
            max_one = max(max_one, one)
            max_zero = max(max_zero, zero)

        return True if max_one>max_zero else False
```
