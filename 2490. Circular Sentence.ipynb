#### Solution 1 in python

```python
class Solution:
    def isCircularSentence(self, sentence: str) -> bool:
        s = sentence.split(" ")
        arr = []
        arr.append(s[0][0])
        for i in range(len(s)-1):
            if s[i][len(s[i])-1]!=s[i+1][0]:
                return False
        arr.append(s[len(s)-1][-1])
        if arr[0] != arr[1]:
            return False
        
        return True
```
