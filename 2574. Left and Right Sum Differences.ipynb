#### Solution 1 in python

```python
class Solution:
    def leftRightDifference(self, nums: List[int]) -> List[int]:
        ls = 0
        rs = 0
        l_sum = [0]
        r_sum = [0]
        for i in range(len(nums)):
            ls+=nums[i]
            l_sum.append(ls)
            rs +=nums[len(nums)-1-i]
            r_sum.append(rs)
        
        l_sum.pop()
        r_sum.pop()
        r_sum = r_sum[::-1]

        ans = []
        for i in range(len(l_sum)):
            ans.append(abs(l_sum[i]-r_sum[i]))

        return ans
```
