```python
class Solution:
    def splitNum(self, num: int) -> int:

        num = list(str(num))
        num.sort()
        num1 = ''.join([f'{num[i]}' for i in range(len(num)) if i%2!=0])
        num2 = "".join([f'{num[i]}' for i in range(len(num)) if i%2==0])
        print(num1, num2)
        ans1 = 0
        ans2 = 0
        for i in num1:
            ans1 = ans1*10 +int(i)

        for i in num2:
            ans2=ans2*10+int(i)


        return ans1+ans2
```
