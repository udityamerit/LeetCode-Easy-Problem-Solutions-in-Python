#### Solution 1 in python

```python
class Solution:
    def reverseVowels(self, s: str) -> str:
        s_list = list(s)
        v = "aeiouAEIOU"
        l=0
        r=len(s)-1
        while l<r:
            if s_list[l] not in v:
                l+=1
            elif s_list[r] not in v:
                r-=1
            else:
                s_list[l], s_list[r]=s_list[r], s_list[l]
                l+=1
                r-=1

        return "".join(s_list)
```
