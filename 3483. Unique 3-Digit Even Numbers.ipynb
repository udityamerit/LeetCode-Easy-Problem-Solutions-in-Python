#### Solution 1 in python

```python
from itertools import permutations
class Solution:
    def totalNumbers(self, digits: List[int]) -> int:
        l =[]
        ans = []
        for i in list(permutations(digits, 3)):
            if i not in l:
                result = 0
                for digit in list(i):
                    result = result * 10 + digit
                
                if (result not in ans and result%2==0):
                    c = len(str(result))

                    if c>2: 
                        ans.append(result)

        return len(ans)
```
