#### Solution 1 in python

```python
class Solution:
    def checkDivisibility(self, n: int) -> bool:
        def sumprod(n):
            su = 0
            p = 1
            for _ in range(len(str(n))):
                s = n%10
                su+=s
                p*=s
                n=n//10
            return su, p
        su, p = sumprod(n)
        
        if n%(su+p)==0:
            return True
        else:
            return False
        
```
