#### Solution 1 in python

```python
class Solution:
    def gcdOfOddEvenSums(self, n: int) -> int:
        even = 0
        odd = 1
        e = []
        o=[]
        for i in range(n):
            even+=2
            e.append(even)
            o.append(odd)
            odd+=2
            

        # print(e, o)
        se = sum(e)
        so = sum(o)

        while so:
            se, so = so, se%so
        return se
```
