#### Solution 1 in python

```python
class Solution:
    def findMissingElements(self, nums: List[int]) -> List[int]:
        s = []
        low = min(nums)
        high = max(nums)
        n  = [i for i in range(low, high+1)]
        for i in n:
            if i not in nums:
                s.append(i)

        return s
```
