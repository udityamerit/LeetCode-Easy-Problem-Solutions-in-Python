```python
class Solution:
    def pivotIndex(self, nums: List[int]) -> int:
        # su = sum(nums)
        # lsum = 0
        # rsum = 0
        # for i in range(1,len(nums)):
        #     lsum+=nums[i-1]
        #     rsum+=nums[-i]
        #     print(lsum, su-rsum)
        #     if lsum==su-rsum:
        #         return i
        
        # return -1

        total_sum = sum(nums)
        left_sum = 0
        
        for i, num in enumerate(nums):
            # Right sum is total minus left sum minus the current element
            if left_sum == (total_sum - left_sum - num):
                return i
            left_sum += num
            
        return -1
```
