#### Solution 1 in python

```python
class Solution:
    def reverseOnlyLetters(self, s: str) -> str:
       
        s_list = list(s)
        
        l = 0
        r = len(s) - 1  
        
        while l < r:
            if not s_list[l].isalpha():
                l += 1  
            elif not s_list[r].isalpha():
                r -= 1  
            else:
               
                s_list[l], s_list[r] = s_list[r], s_list[l]
                l += 1
                r -= 1  
                
        return "".join(s_list)  
```
